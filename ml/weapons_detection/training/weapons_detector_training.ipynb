```python
data_yaml_content = """train: train/images
val: valid/images
test: test/images

nc: 7
names: ['Handgun', 'Knife', 'Missile', 'Rifle', 'Shotgun', 'Sword', 'Tank']
"""

with open("weapons_dataset/data.yaml", "w") as f:
    f.write(data_yaml_content)

print("data.yaml fixed!")
```

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

results = model.train(
    data="weapons_dataset/data.yaml",
    epochs=50,
    imgsz=640,
    batch=16,
    name="weapons_model"
)
```
